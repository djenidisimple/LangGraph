```python
from typing from TypedDict, List
from langgraph.graph import StateGraph
import math
```

```python
class AgentState(TypedDict):
    values: List[int]
    name: str
    operation: str
    result: str
```

```python
def process_values(state: AgentState) -> AgentState:
    """"Node function that handles multiple different inputs"""
    if state['operation'] == '+':
        state['result'] = f"Hi there {state['name']}, your answer is : {sum(state['values'])}"
    elif state['operation'] == '*':
        state['result'] = f"Hi there {state['name']}, your answer is : {math.prod(state['values'])}"
    else:
        state['result'] = f"Hi there {state['name']}, your operation is not supported"
    return state  
```

```python
graph = StateGraph(AgentState)
graph.add_node("processor", process_values) # Add a node to graph
graph.set_entry_point("processor") # set the entry point of the graph to the node we just added
graph.set_finish_point("processor") # set the finish point of the graph to the node we

app = graph.compile()
```

```python
from IPython.display import display, Image
display(Image(app.get_graph().draw_mermaid_png()))
```

```python
result = app.invoke({"name" : "Jack Sparrow", "values" : [1, 2, 3, 4], "operation" : "*"})

result['result']
```
